```python
# 1. Khai báo thư viện
import pandas as pd
import numpy as np
# 2. Nạp dữ liệu TikTok Shop vào DataFrame
df = pd.read_csv('tiktok_shop_sales.csv')
# 3. Xem kích thước (Số dòng, Số cột) của dữ liệu
print(f"Tổng số dòng: {df.shape[0]}")
print(f"Tổng số cột: {df.shape[1]}")
# 4. Hiển thị 5 dòng dữ liệu đầu tiên
df.head()
```

```python
# Kiểm tra thông tin cấu trúc và kiểu dữ liệu
df.info()
```

```python
# Thống kê số lượng giá trị khuyết thiếu
print(" 1. SỐ LƯỢNG GIÁ TRỊ NULL TRÊN TỪNG CỘT ")
print(df.isnull().sum())

# 2.Kiểm tra số lượng dòng bị trùng lặp hoàn toàn
duplicates_count = df.duplicated().sum()
print(f"\n 2. TỔNG SỐ DÒNG BỊ TRÙNG LẶP HOÀN TOÀN: {duplicates_count} dòng")
```

```python
# Kiểm tra các giá trị khác biệt trong cột Người sáng tạo / KOC (Creator_Name)
print(" 1. CÁC GIÁ TRỊ KHÁC BIỆT CỦA CỘT Creator_Name ")
print(df['Creator_Name'].value_counts().head(8))

# Kiểm tra các giá trị khác biệt của Tỉnh / Thành phố (Shipping_City)
print("\n 2. DANH SÁCH TỈNH / THÀNH PHỐ (Shipping_City) ---")
print(df['Shipping_City'].unique())

# Kiểm tra các trạng thái đơn hàng (Order_Status)
print("\n3. TRẠNG THÁI ĐƠN HÀNG (Order_Status)")
print(df['Order_Status'].value_counts())
```

```python
# Tạo bản sao df_clean để làm sạch
df_clean = df.copy()

# Xóa cột thừa không phục vụ phân tích đơn hàng (cột Is_Anomaly)
if 'Is_Anomaly' in df_clean.columns:
    df_clean.drop(columns=['Is_Anomaly'], inplace=True)
    print("-> [XONG] Đã xóa cột thừa: Is_Anomaly")

# Chuyển đổi cột Order_Date từ kiểu chữ (object) sang đúng kiểu Ngày Giờ (datetime)
df_clean['Order_Date'] = pd.to_datetime(df_clean['Order_Date'])
print(f"-> [XONG] Kiểu dữ liệu mới của Order_Date: {df_clean['Order_Date'].dtype}")

# So sánh số lượng cột Trước (Before) và Sau (After)
print(f"\n SO SÁNH SỐ CỘT")
print(f"Trước khi làm sạch (Before): {df.shape[1]} cột")
print(f"Sau khi làm sạch (After) : {df_clean.shape[1]} cột")
```

```python
# Chuẩn hóa giá trị khuyết thiếu ngầm ở cột Creator_Name
df_clean['Creator_Name'] = df_clean['Creator_Name'].replace(
    'None (Gian hàng trực tiếp)',
    'Bán trực tiếp (Không qua KOC)')
print("-> [XONG] Đã chuẩn hóa giá trị thiếu ngầm ở cột Creator_Name.")

# Cột số (Customer_Age) nếu có khuyết sẽ điền bằng median (trung vị)
median_age = df_clean['Customer_Age'].median()
df_clean['Customer_Age'] = df_clean['Customer_Age'].fillna(median_age)

# Cột phân loại (Shipping_City) nếu có khuyết sẽ điền bằng mode (giá trị phổ biến nhất)
mode_city = df_clean['Shipping_City'].mode()[0]
df_clean['Shipping_City'] = df_clean['Shipping_City'].fillna(mode_city)
print(f"-> [XONG] Đã cấu hình chiến lược Impute: Age (Median={median_age}), City (Mode='{mode_city}').")

# Loại bỏ dòng rác: Xóa các dòng bị khuyết trường khóa chính Order_ID (nếu có)
rows_before = len(df_clean)
df_clean = df_clean.dropna(subset=['Order_ID'])
df_clean = df_clean.dropna(how='all') # Xóa dòng trống hoàn toàn
print(f"-> [XONG] Kiểm tra khóa chính: Giữ vững {len(df_clean)}/{rows_before} dòng hợp lệ.")

# Hiển thị lại tỷ lệ phân bổ của Creator_Name sau chuẩn hóa
print("\n--- PHÂN BỔ KOC/CREATOR SAU KHI XỬ LÝ KHUYẾT THIẾU ---")
print(df_clean['Creator_Name'].value_counts().head(5))
```

```python
# Loại bỏ trùng lặp hoàn toàn (nếu có)
df_clean = df_clean.drop_duplicates()

# Xử lý trùng lặp một phần theo khóa chính Order_ID:
# Nếu trùng mã đơn hàng, sắp xếp theo thời gian và giữ lại bản ghi mới nhất
df_clean = df_clean.sort_values('Order_Date').drop_duplicates(subset=['Order_ID'], keep='last')
df_clean.reset_index(drop=True, inplace=True)
print(f"-> [XONG] Đảm bảo tính duy nhất tuyệt đối: {df_clean['Order_ID'].nunique()}/{len(df_clean)} đơn hàng.")

# BẢNG TỔNG KẾT ĐỐI CHIẾU TRƯỚC VÀ SAU KHI LÀM SẠCH (BEFORE VS AFTER)
print("\n" + "="*60)
print("       BẢNG ĐỐI CHIẾU KẾT QUẢ TRƯỚC VÀ SAU KHI LÀM SẠCH")
print("="*60)
summary_data = {
    "Tiêu chí": [
        "Tổng số dòng (Rows)",
        "Tổng số cột thuộc tính (Columns)",
        "Kiểu dữ liệu cột Order_Date",
        "Số giá trị khuyết ngầm Creator_Name",
        "Cột rác / thừa (Is_Anomaly)",
        "Tính duy nhất của Order_ID"
    ],
    "Trước khi làm sạch (Before)": [
        f"{df.shape[0]} dòng",
        f"{df.shape[1]} cột",
        str(df['Order_Date'].dtype) + " (Chuỗi ký tự/object)",
        f"{df['Creator_Name'].str.contains('None', na=False).sum()} dòng ('None')",
        "Có tồn tại (1 cột rác)",
        "Chưa kiểm soát xung đột"
    ],
    "Sau khi làm sạch (After)": [
        f"{df_clean.shape[0]} dòng",
        f"{df_clean.shape[1]} cột (Đã tinh gọn)",
        str(df_clean['Order_Date'].dtype) + " (Chuẩn thời gian)",
        f"{df_clean['Creator_Name'].str.contains('None', na=False).sum()} dòng (Đã gán nhãn chuẩn)",
        "Đã loại bỏ hoàn toàn (0 cột)",
        "100% Duy nhất (Unique)"]}
df_summary = pd.DataFrame(summary_data)
display(df_summary)
```
